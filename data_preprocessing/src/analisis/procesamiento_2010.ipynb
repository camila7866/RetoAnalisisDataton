```python
import pandas as pd
```

**Obtención de poblacion mayor a 60 años por alcaldía**

```python
pob_2010 = pd.read_excel("../../data/raw/2010/pob_cpv2010_cdmx.xls", header=6, usecols=["Entidad federativa", "Delegación", "Sexo","Grupos quinquenales de edad", "Población total1"])
pob_2010.head(5)
```

```python
pob_2010.drop(index=[0,1], inplace=True)
```

```python
pob_2010.rename(columns={"Entidad federativa": "entidad", "Delegación": "mun", "Sexo":"sexo", "Grupos quinquenales de edad": "rangos_edad", "Población total1": "poblacion"}, inplace=True)
```

```python
pob_2010 = pob_2010[(pob_2010["mun"] != "Total") & (pob_2010["sexo"] == "Total")]
pob_2010.sample(5)
```

```python
rangos_edad = ["60-64 años", "65-69 años", "70-74 años", "75-79 años", "80-84 años", "85-89 años", "90-94 años", "95-99 años", "100 años y más"]
```

```python
pob_2010 = pob_2010[pob_2010["rangos_edad"].isin(rangos_edad)]
pob_2010.head(5)
```

```python
pob_2010["entidad"] = pob_2010["entidad"].str.extract("(\d*)\s")
pob_2010["mun"] = pob_2010["mun"].str.extract("(\d*)\s")
pob_2010.head(5)
```

```python
pob_2010_60mas = pob_2010.groupby(["entidad","mun"], as_index=False)["poblacion"].sum()
pob_2010_60mas.rename(columns={"poblacion": "p_60ymas_mun"}, inplace=True)
pob_2010_60mas
```

**Obtención de población mayor a 60 años por AGEB**

Como podemos ver los campos que utilizaremos seran:
* Clave entidad federativa - ENTIDAD
* Entidad federativa - NOM_ENT
* Clave de municipio o demarcación territorial - MUN
* Municipio o demarcación territorial - NOM_MUN
* Clave de localidad - LOC
* Localidad	- NOM_LOC
* Clave del AGEB - AGEB
* Clave manzana - MZA
* Población total - POBTOT
* Población de 60 años y más - P_60YMAS

```python
pob_ageb_2010 = pd.read_csv("../../data/raw/2010/pob_ageb_cpv2010_cdmx.csv", dtype={"entidad":str, "mun": str, "loc": str})
pob_ageb_2010
```

```python
pob_ageb_2010 = pob_ageb_2010[["entidad", "nom_ent", "mun", "nom_mun", "loc", "nom_loc", "ageb", "mza", "pobtot", "p_60ymas"]]
pob_ageb_2010.info()
```

```python
pob_ageb_2010["entidad"] = pob_ageb_2010["entidad"].str.zfill(2)
pob_ageb_2010["mun"] = pob_ageb_2010["mun"].str.zfill(3)
pob_ageb_2010["loc"] = pob_ageb_2010["loc"].str.zfill(4)
pob_ageb_2010["ageb"] = pob_ageb_2010["ageb"].str.zfill(4)
```

```python
pob_ageb_2010 = pob_ageb_2010[(pob_ageb_2010.ageb != "0000") & (pob_ageb_2010.mza == 0)]
pob_ageb_2010.sample(5)
```

Pasemos p_60ymas a numerico para poder operarla.

```python
pob_ageb_2010["p_60ymas"] = pd.to_numeric(pob_ageb_2010["p_60ymas"], errors="coerce")
```

```python
pob_ageb_2010[pob_ageb_2010["p_60ymas"].isna()]
```

Antes de ver que haremos con esos NaN haremos la clave unica geoestadítica por ageb

```python
pob_ageb_2010["cvegeo"] = (pob_ageb_2010["entidad"] + pob_ageb_2010["mun"] + pob_ageb_2010["loc"] + pob_ageb_2010["ageb"])
```

```python
pob_ageb_2010.sample(5)
```

**Obtención de servicios DENUE por AGEB**

*Servicios DENUE 2010* 
Clave - Descripción de categoría
* 621610 - Servicios de enfermería a domicilio
* 623311 - Asilos y otras residencias del sector privado para el cuidado de ancianos
* 623312 - Asilos y otras residencias del sector público para el cuidado de ancianos

```python
denue_2010 = pd.read_csv("../../data/raw/2010/denue2010.csv", encoding="latin-1")
denue_2010.info()
```

```python
denue_2010.sample()
```

Lo que unicamente utilizaremos sera:
* Código de la clase de actividad
* Entidad federativa
* Municipio 
* Localidad
* Área geoestadística básica

```python
denue_2010 = denue_2010[["Código de la clase de actividad", "Entidad federativa", "Municipio", "Localidad", "Área geoestadística básica"]]
```

```python
denue_2010["Entidad federativa"] = "09"
```

```python
denue_2010["Municipio"] = denue_2010["Municipio"].str.lower()
denue_2010["Localidad"] = denue_2010["Localidad"].str.lower()
denue_2010.sample()
```

Carguemos un listado de las alcaldia y sus localidades ya hecho.

```python
lista_loc = pd.read_csv("../../data/raw/2010/localidades_2010.csv", dtype={"ent": str, "mun": str, "loc": str})
lista_loc["nom_mun"] = lista_loc["nom_mun"].str.lower()
lista_loc["nom_loc"] = lista_loc["nom_loc"].str.lower()
lista_loc
```

```python
lista_loc.info()
```

```python
merge_denue = denue_2010.merge(lista_loc, left_on=["Entidad federativa", "Municipio", "Localidad"] ,right_on=["ent", "nom_mun", "nom_loc"], how="left")
```

```python
merge_denue[merge_denue["nom_loc"].isna()]["Localidad"]
```

Hay localidades no identificadas, pero esperemos a ver si dentro de nuestras actividades siguen habiendo.

```python
merge_denue.rename(columns={"Código de la clase de actividad": "codigo_act", "Área geoestadística básica": "ageb"}, inplace=True)
```

```python
denue_2010 = merge_denue[["codigo_act", "ent", "mun", "nom_mun", "loc", "nom_loc", "ageb"]]
```

```python
denue_2010["cvegeo"] = (denue_2010["ent"] + denue_2010["mun"] + denue_2010["loc"] + denue_2010["ageb"])
denue_2010.sample()
```

```python
serv_enf_dom = denue_2010[denue_2010["codigo_act"] == 621610]
```

```python
serv_enf_dom = serv_enf_dom.groupby("cvegeo", as_index=False)["codigo_act"].count()
serv_enf_dom.rename(columns={"codigo_act":"cantidad_serv_enf_dom"}, inplace=True)
serv_enf_dom
```

```python
asilos = denue_2010[denue_2010["codigo_act"].isin([623311, 623312])]
asilos
```

```python
asilos = asilos.groupby("cvegeo", as_index=False)["codigo_act"].count()
asilos.rename(columns={"codigo_act":"cantidad_asilos"}, inplace=True)
asilos
```

Como de las agebs que no tienen establecimientos, es porque no hay información en el denue, entonces se infiere que no hay esos servicios, por lo que en esas rellenaremos con 0.

```python
res = pob_ageb_2010.merge(serv_enf_dom, on="cvegeo", how="left")
res["cantidad_serv_enf_dom"] = res["cantidad_serv_enf_dom"].fillna(0)
```

```python
res
```

```python
res = res.merge(asilos, on="cvegeo", how="left")
res["cantidad_asilos"] = res["cantidad_asilos"].fillna(0)
res
```

Saquemos la concentración de poblacion adulta mayor por ageb para la estimación de 2015.

```python
res = res.merge(pob_2010_60mas[["mun", "p_60ymas_mun"]], on="mun", how="left")
```

```python
res["concentracion"] = res["p_60ymas"] / res["p_60ymas_mun"]
```

```python
res.to_csv("../../data/processed/results2010.csv") 
```
