```python
import pandas as pd
```

**Obtención de poblacion mayor a 60 años por alcaldía**

```python
pob_2020 = pd.read_excel("../../data/raw/2020/pob_cpv2020_cdmx.xlsx", sheet_name="02", header=5, usecols=["Entidad federativa", "Demarcación territorial", "Sexo","Grupos quinquenales de edad", "Población total1"])
pob_2020.head(5)
```

```python
pob_2020.drop(index=[0,1], inplace=True)
```

```python
pob_2020.rename(columns={"Entidad federativa": "entidad", "Demarcación territorial": "mun", "Sexo":"sexo", "Grupos quinquenales de edad": "rangos_edad", "Población total1": "poblacion"}, inplace=True)
```

```python
pob_2020 = pob_2020[(pob_2020["mun"] != "Total") & (pob_2020["sexo"] == "Total")]
pob_2020.sample(5)
```

```python
rangos_edad = ["60-64 años", "65-69 años", "70-74 años", "75-79 años", "80-84 años", "85-89 años", "90-94 años", "95-99 años", "100 años y más"]
```

```python
pob_2020 = pob_2020[pob_2020["rangos_edad"].isin(rangos_edad)]
pob_2020.head(5)
```

```python
pob_2020["entidad"] = pob_2020["entidad"].str.extract("(\d*)\s")
pob_2020["mun"] = pob_2020["mun"].str.extract("(\d*)\s")
pob_2020.head(5)
```

```python
pob_2020_60mas = pob_2020.groupby(["entidad","mun"], as_index=False)["poblacion"].sum()
pob_2020_60mas.rename(columns={"poblacion": "p_60ymas_mun"}, inplace=True)
pob_2020_60mas
```

**Obtención de población mayor a 60 años por AGEB**

Como podemos ver los campos que utilizaremos seran:
* Clave entidad federativa - ENTIDAD
* Entidad federativa - NOM_ENT
* Clave de municipio o demarcación territorial - MUN
* Municipio o demarcación territorial - NOM_MUN
* Clave de localidad - LOC
* Localidad	- NOM_LOC
* Clave del AGEB - AGEB
* Clave manzana - MZA
* Población total - POBTOT
* Población de 60 años y más - P_60YMAS

```python
pob_ageb_2020 = pd.read_csv("../../data/raw/2020/pob_ageb_cpv2020_cdmx.csv", dtype={"ENTIDAD":str, "MUN": str, "LOC": str})
pob_ageb_2020
```

```python
pob_ageb_2020.rename(columns={"ENTIDAD": "entidad", "NOM_ENT": "nom_ent", "MUN": "mun", "NOM_MUN": "nom_mun", "LOC": "loc", "NOM_LOC": "nom_loc", "AGEB": "ageb", "MZA": "mza", "POBTOT": "pobtot", "P_60YMAS": "p_60ymas"}, inplace=True)
```

```python
pob_ageb_2020 = pob_ageb_2020[["entidad", "nom_ent", "mun", "nom_mun", "loc", "nom_loc", "ageb", "mza", "pobtot", "p_60ymas"]]
pob_ageb_2020.info()
```

```python
pob_ageb_2020["entidad"] = pob_ageb_2020["entidad"].str.zfill(2)
pob_ageb_2020["mun"] = pob_ageb_2020["mun"].str.zfill(3)
pob_ageb_2020["loc"] = pob_ageb_2020["loc"].str.zfill(4)
pob_ageb_2020["ageb"] = pob_ageb_2020["ageb"].str.zfill(4)
```

```python
pob_ageb_2020 = pob_ageb_2020[(pob_ageb_2020.ageb != "0000") & (pob_ageb_2020.mza == 0)]
pob_ageb_2020.sample(5)
```

Pasemos p_60ymas a numerico para poder operarla.

```python
pob_ageb_2020["p_60ymas"] = pd.to_numeric(pob_ageb_2020["p_60ymas"], errors="coerce")
```

```python
pob_ageb_2020[pob_ageb_2020["p_60ymas"].isna()]
```

Antes de ver que haremos con esos NaN haremos la clave unica geoestadítica por ageb

```python
pob_ageb_2020["cvegeo"] = (pob_ageb_2020["entidad"] + pob_ageb_2020["mun"] + pob_ageb_2020["loc"] + pob_ageb_2020["ageb"])
```

```python
pob_ageb_2020.sample(5)
```

**Obtención de servicios DENUE por AGEB**

*Servicios DENUE 2020* 
Clave - Descripción de categoría
* 621610 - Servicios de enfermería a domicilio
* 623311 - Asilos y otras residencias del sector privado para el cuidado de ancianos
* 623312 - Asilos y otras residencias del sector público para el cuidado de ancianos

```python
denue_2020 = pd.read_csv("../../data/raw/2020/denue2020.csv", encoding="latin-1")
denue_2020.info()
```

```python
denue_2020.sample()
```

Lo que unicamente utilizaremos sera:
* Código de la clase de actividad
* Entidad federativa
* Municipio 
* Localidad
* Área geoestadística básica

```python
denue_2020 = denue_2020[["codigo_act", "cve_ent", "cve_mun", "cve_loc", "ageb"]]
denue_2020.info()
```

Hay localidades no identificadas, pero esperemos a ver si dentro de nuestras actividades siguen habiendo.

```python
denue_2020.rename(columns={"cve_ent":"entidad", "cve_mun":"mun", "cve_loc":"loc"}, inplace=True)
```

```python
denue_2020["entidad"] = denue_2020["entidad"].astype(str).str.zfill(2)
denue_2020["mun"] = denue_2020["mun"].astype(str).str.zfill(3)
denue_2020["loc"] = denue_2020["loc"].astype(str).str.zfill(4)
denue_2020["ageb"] = denue_2020["ageb"].str.zfill(4)
denue_2020.sample()
```

```python
denue_2020["cvegeo"] = (denue_2020["entidad"] + denue_2020["mun"] + denue_2020["loc"] + denue_2020["ageb"])
denue_2020.sample()
```

```python
serv_enf_dom = denue_2020[denue_2020["codigo_act"] == 621610]
```

```python
serv_enf_dom = serv_enf_dom.groupby("cvegeo", as_index=False)["codigo_act"].count()
serv_enf_dom.rename(columns={"codigo_act":"cantidad_serv_enf_dom"}, inplace=True)
serv_enf_dom
```

```python
asilos = denue_2020[denue_2020["codigo_act"].isin([623311, 623312])]
asilos
```

```python
asilos = asilos.groupby("cvegeo", as_index=False)["codigo_act"].count()
asilos.rename(columns={"codigo_act":"cantidad_asilos"}, inplace=True)
asilos
```

Como de las agebs que no tienen establecimientos, es porque no hay información en el denue, entonces se infiere que no hay esos servicios, por lo que en esas rellenaremos con 0.

```python
res = pob_ageb_2020.merge(serv_enf_dom, on="cvegeo", how="left")
res["cantidad_serv_enf_dom"] = res["cantidad_serv_enf_dom"].fillna(0)
```

```python
res
```

```python
res = res.merge(asilos, on="cvegeo", how="left")
res["cantidad_asilos"] = res["cantidad_asilos"].fillna(0)
res
```

Saquemos la concentración de poblacion adulta mayor por ageb para la estimación de 2015.

```python
res = res.merge(pob_2020_60mas[["mun", "p_60ymas_mun"]], on="mun", how="left")
```

```python
res["concentracion"] = res["p_60ymas"] / res["p_60ymas_mun"]
```

```python
res.sample(5)
```
