**Obtencion de población mayor a 60 años por alcaldía**

```python
import pandas as pd
```

```python
pob_2015 = pd.read_excel("../../data/raw/2015/pob_eic2015_cdmx.xls", sheet_name="02", header=6)
pob_2015.head()
```

```python
pob_2015.drop(index=0, inplace=True)
pob_2015.head()
```

```python
pob_2015.rename(columns={"Entidad federativa": "entidad", "Delegación": "mun", "Grupos quinquenales de edad": "rangos_edad", "Población total": "poblacion"}, inplace=True)
```

```python
rangos_edad = ["60-64 años", "65-69 años", "70-74 años", "75 años y más"]
```

```python
pob_2015 = pob_2015[(pob_2015["mun"] != "Total") & (pob_2015["rangos_edad"].isin(rangos_edad)) & (pob_2015["Estimador"] == "Valor")]
pob_2015
```

```python
pob_2015 = pob_2015[["entidad","mun","rangos_edad","poblacion"]]
pob_2015
```

```python
pob_2015["entidad"] = pob_2015["entidad"].str.extract("(\d*)\s")
pob_2015["mun"] = pob_2015["mun"].str.extract("(\d*)\s")
pob_2015.head(5)
```

```python
pob_2015_60mas = pob_2015.groupby(["entidad","mun"], as_index=False)["poblacion"].sum()
pob_2015_60mas.rename(columns={"poblacion": "p_60ymas_mun"}, inplace=True)
pob_2015_60mas
```

Veamos que AGEBs estan presentes en un año y no en el otro. Las eliminaremos ya que utilizaremos el mapa de 2020.

```python
a = r2010[["entidad","cvegeo","concentracion"]].merge(r2020[["entidad","cvegeo","concentracion"]], on="cvegeo", how="outer")
```

```python
a[a["entidad_y"].isna()]
```

```python
a.drop(index=[515,2240], inplace=True)
```

```python
a["concentracion_prom"] = (a["concentracion_x"] + a["concentracion_y"])/2
```

```python
import numpy as np
```

AGEBs totalmente confidenciales

```python
a.loc[(a["concentracion_x"].isna()) & (a["concentracion_y"].isna()),"concentracion_prom"] = np.nan
```

```python
a.loc[(a["concentracion_x"].isna()) & (a["concentracion_y"].notna()), "concentracion_prom"] = a["concentracion_y"]
```

```python
a.loc[(a["concentracion_x"].notna()) & (a["concentracion_y"].isna()), "concentracion_prom"] = a["concentracion_x"]
```

```python
a = a[["cvegeo","concentracion_prom"]]
a.sample(5)
```

Tomamos la lista de AGEB 2020

```python
merge2020 = r2020[["entidad", "nom_ent", "mun", "nom_mun", "loc", "nom_loc", "ageb", "cvegeo"]].merge(a, on="cvegeo", how="left")
```

```python
merge2020.sample()
```

```python
merge2020 = merge2020.merge(pob_2015_60mas[["mun", "p_60ymas_mun"]], on="mun", how="left")
```

```python
merge2020[merge2020["concentracion_prom"].isna()]
```

```python
merge2020["p_60ymas_est"] = merge2020["p_60ymas_mun"] * merge2020["concentracion_prom"]
merge2020.sample()
```

```python
pob_ageb_2015 = merge2020[["entidad", "nom_ent", "mun", "nom_mun", "loc", "nom_loc", "ageb", "cvegeo", "p_60ymas_est"]]
pob_ageb_2015.sample()
```

**Obtención de servicios DENUE por AGEB**

*Servicios DENUE 2015* 
Clave - Descripción de categoría
* 621610 - Servicios de enfermería a domicilio
* 623311 - Asilos y otras residencias del sector privado para el cuidado de ancianos
* 623312 - Asilos y otras residencias del sector público para el cuidado de ancianos

```python
denue_2015 = pd.read_csv("../../data/raw/2015/denue2015.csv")
denue_2015.info()
```

```python
denue_2015 = denue_2015[["Código de la clase de actividad SCIAN", "Clave entidad", "Clave municipio", "Clave localidad", "Área geoestadística básica "]]
denue_2015.info()
```

```python
denue_2015.rename(columns={"Código de la clase de actividad SCIAN": "codigo_act", "Clave entidad": "entidad", "Clave municipio": "mun", "Clave localidad": "loc", "Área geoestadística básica ": "ageb"}, inplace=True)
```

```python
denue_2015["entidad"] = denue_2015["entidad"].astype(str).str.zfill(2)
denue_2015["mun"] = denue_2015["mun"].astype(str).str.zfill(3)
denue_2015["loc"] = denue_2015["loc"].astype(str).str.zfill(4)
denue_2015["ageb"] = denue_2015["ageb"].astype(str).str.zfill(4)
denue_2015.sample()
```

```python
denue_2015["cvegeo"] = (denue_2015["entidad"] + denue_2015["mun"] + denue_2015["loc"] + denue_2015["ageb"])
denue_2015.sample()
```

```python
serv_enf_dom = denue_2015[denue_2015["codigo_act"] == 621610]
```

```python
serv_enf_dom = serv_enf_dom.groupby("cvegeo", as_index=False)["codigo_act"].count()
serv_enf_dom.rename(columns={"codigo_act":"cantidad_serv_enf_dom"}, inplace=True)
serv_enf_dom
```

```python
asilos = denue_2015[denue_2015["codigo_act"].isin([623311, 623312])]
asilos
```

```python
asilos = asilos.groupby("cvegeo", as_index=False)["codigo_act"].count()
asilos.rename(columns={"codigo_act":"cantidad_asilos"}, inplace=True)
asilos
```

Como de las agebs que no tienen establecimientos, es porque no hay información en el denue, entonces se infiere que no hay esos servicios, por lo que en esas rellenaremos con 0.

```python
res = pob_ageb_2015.merge(serv_enf_dom, on="cvegeo", how="left")
res["cantidad_serv_enf_dom"] = res["cantidad_serv_enf_dom"].fillna(0)
```

```python
res = res.merge(asilos, on="cvegeo", how="left")
res["cantidad_asilos"] = res["cantidad_asilos"].fillna(0)
res
```

```python
res.to_csv("../../data/processed/results2015.csv") 
```
